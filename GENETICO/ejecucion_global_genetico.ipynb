Este notebook ejecuta el algoritmo para cada CSV de `INSTANCIAS`, con las semillas 1 a 5 y 500 generaciones por corrida.
```python
from pathlib import Path
import sys

# Encuentra ga_tournament.py si Jupyter se inició desde GENETICO o desde el proyecto.
_candidatos = (
    Path.cwd(),
    Path.cwd() / "GENETICO",
    Path.cwd() / "entrega2" / "GENETICO",
    Path.cwd().parent / "GENETICO",
)
GENETICO_DIR = next(
    (ruta for ruta in _candidatos if (ruta / "ga_tournament.py").is_file()),
    None,
)
if GENETICO_DIR is None:
    raise FileNotFoundError(
        "No se encontró ga_tournament.py. Abra el notebook desde la carpeta "
        "GENETICO o mantenga ambos archivos en la misma carpeta."
    )
if str(GENETICO_DIR) not in sys.path:
    sys.path.insert(0, str(GENETICO_DIR))

import importlib
import ga_tournament
ga_tournament = importlib.reload(ga_tournament)
find_instances_dir = ga_tournament.find_instances_dir
mutation_rate_for_size = ga_tournament.mutation_rate_for_size
run_all_instances = ga_tournament.run_all_instances

INSTANCIAS_DIR = find_instances_dir()
SEMILLAS = range(1, 6)
GENERACIONES = 500
PESO_BREAKS = 10
PESO_OTRAS_LIGAS = 4

print(f"Código genético: {GENETICO_DIR}")
print(f"Instancias: {INSTANCIAS_DIR}")
print(f"Semillas: {list(SEMILLAS)} | generaciones por ejecución: {GENERACIONES}")

print("Mutaci?n por tama?o:", {equipos: mutation_rate_for_size(equipos) for equipos in (8, 10, 12, 14, 16)})
```
