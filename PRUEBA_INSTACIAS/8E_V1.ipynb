# Prueba de instancia: 8 equipos, versi?n 1

Este notebook resuelve `Instancia_Sintetica_8Eq_v1.csv` con el modelo matem?tico compartido. Al ejecutar la celda se muestra el estado, el valor objetivo y el calendario; si existe una soluci?n factible, tambi?n se guarda el calendario en `RESULTADOS`.

```python
from pathlib import Path
import sys
import pandas as pd

_candidatos = (Path.cwd(), Path.cwd().parent, Path.cwd() / "entrega2")
ROOT = next((p for p in _candidatos if (p / "round_robin_gurobi_modelo.py").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("No se encontró round_robin_gurobi_modelo.py desde la carpeta actual ni su padre.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from round_robin_gurobi_ejecucion import guardar_calendario, resolver_instancia

ruta_csv = ROOT / "INSTANCIAS" / "Instancia_Sintetica_8Eq_v1.csv"
ruta_resultados = ROOT / "RESULTADOS"

resultado = resolver_instancia(
    ruta_csv,
    w_b=1.0,
    w_ext=2.0,
    time_limit=3600,
    mip_gap=0.0,
    verbose=False,
)
ruta_calendario = guardar_calendario(resultado, ruta_resultados)
print({clave: valor for clave, valor in resultado.items() if clave != "Calendario"})
if ruta_calendario is not None:
    print(f"Calendario guardado en: {ruta_calendario}")

pd.DataFrame(resultado["Calendario"])
```
